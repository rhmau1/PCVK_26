1. Apa yang dimaksud dengan piksel dalam citra digital?

Jika citra yang dibaca dengan OpenCV langsung ditampilkan lewat Matplotlib tanpa konversi warna terlebih dahulu, tampilan warna citra akan tertukar (misalnya warna kulit tampak kebiruan).

5. Mengapa Grad-CAM++ penting dalam visi komputer modern?
- Grad-CAM++ (Generalized Gradient-weighted Class Activation Mapping) berperan krusial dalam bidang Explainable AI (XAI). Metode ini memvisualisasikan area piksel mana pada citra yang menjadi dasar jaringan syaraf konvolusional (CNN) dalam mengambil keputusan klasifikasi. Dibandingkan Grad-CAM standar, Grad-CAM++ memberikan bobot gradien orde kedua dan ketiga yang lebih baik sehingga mampu melokalisasi beberapa objek dari kelas yang sama secara bersamaan dan menghasilkan peta panas (heatmap) yang jauh lebih akurat pada detail objek kecil.
6. Apa keunggulan format PNG dibanding JPEG?
- Kompresi Lossless: Menjaga kualitas visual tetap utuh 100% tanpa distorsi kompresi atau artefak buram saat disimpan berulang kali.

- Dukungan Kanal Transparansi (Alpha Channel): Mampu menyimpan latar belakang transparan atau semi-transparan, fitur yang tidak dimiliki oleh format JPEG.
7. Mengapa HSV lebih cocok untuk segmentasi warna dibanding RGB?
- Ruang warna RGB menggabungkan informasi warna dan intensitas cahaya ke dalam tiap kanal (R, G, B), sehingga perubahan bayangan atau pencahayaan lingkungan akan mengubah nilai ketiga kanal secara drastis. Sebaliknya, HSV memisahkan corak warna murni pada kanal Hue (H) dari tingkat kejenuhan (Saturation) dan intensitas cahaya (Value). Hal ini membuat proses ambang batas (thresholding) warna objek tertentu menjadi jauh lebih stabil terhadap variasi pencahayaan.

9. Apa itu super-resolution berbasis AI?
- Super-resolution berbasis AI adalah teknik peningkatan dimensi dan resolusi spasial citra dari resolusi rendah (low-resolution) ke resolusi tinggi (high-resolution) menggunakan arsitektur deep learning (seperti CNN, GAN, atau Model Difusi). Alih-alih sekadar menduplikasi atau menginterpolasi piksel secara matematis biasa (seperti bicubic interpolation), AI memprediksi tekstur, tepi, dan detail halus yang hilang berdasarkan pola representasi visual yang telah dipelajari dari dataset latih.
