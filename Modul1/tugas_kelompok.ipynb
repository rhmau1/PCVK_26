```python
from google.colab import files
import cv2 as cv
from google.colab.patches import cv2_imshow
import matplotlib.pyplot as plt

uploaded = files.upload()
img = cv.imread('ktm scan.jpeg')
plt.imshow(img)
plt.show()
```

```python
# Kelompok 1 bagian NIM, nama dan Prodi
kel1 = img.copy()
cv.rectangle(kel1, pt1=(290,650), pt2=(520,690), color=(0,255,0), thickness=-1) #sensor NIM
cv.rectangle(kel1, pt1=(290,610), pt2=(820,640), color=(255,255,0), thickness=-1) #sensor Nama
cv.rectangle(kel1, pt1=(290,700), pt2=(750,740), color=(0,255,255), thickness=-1) #sensor Prodi

plt.imshow(cv.cvtColor(kel1, cv.COLOR_BGR2RGB))
plt.show()
```

```python
# Kelompok 2 bagian mata dan mulut pada Foto Mahasiswa
kel2 = img.copy()

cv.circle(kel2, center=(150,530), radius=15, color=(0,255,0), thickness=3) #sensor mata
cv.circle(kel2, center=(185,530), radius=15, color=(255,255,0), thickness=3) #sensor mata
cv.rectangle(kel2, pt1=(140,555), pt2=(200,575), color=(255,0,0), thickness=2) #sensor mulut

plt.imshow(cv.cvtColor(kel2, cv.COLOR_BGR2RGB))
plt.show()
```

```python
# Kelompok 3 bagian tulisan POLITEKNIK NEGERI MALANG, BNI dan Prodi
kel3 = img.copy()
cv.rectangle(kel3, pt1=(195,30), pt2=(510,125), color=(0,255,0), thickness=-1) #sensor polinema
cv.rectangle(kel3, pt1=(1020,50), pt2=(1210,130), color=(255,255,0), thickness=-1) #sensor BNI
cv.rectangle(kel3, pt1=(290,700), pt2=(750,740), color=(0,255,255), thickness=-1) #sensor Prodi

plt.imshow(cv.cvtColor(kel3, cv.COLOR_BGR2RGB))
plt.show()
```

```python
# Kelompok 4 tulisan Malang, POLINEMA dan Barcode
kel4 = img.copy()
cv.rectangle(kel4, pt1=(195,30), pt2=(510,125), color=(0,255,0), thickness=3) #sensor polinema
cv.rectangle(kel4, pt1=(340,90), pt2=(510,125), color=(255,255,0), thickness=2) #sensor malang
cv.rectangle(kel4, pt1=(1035,420), pt2=(1240,630), color=(0,255,255), thickness=5) #sensor barcode

plt.imshow(cv.cvtColor(kel4, cv.COLOR_BGR2RGB))
plt.show()
```

```python
# Kelompok 5 bagian Logo POLINEMA, tgl Valid Thru dan tulisan D-IV
kel5 = img.copy()
cv.rectangle(kel5, pt1=(75,20), pt2=(180,125), color=(0,255,0), thickness=2) #sensor logo polinema
cv.rectangle(kel5, pt1=(725,500), pt2=(890,550), color=(255,255,0), thickness=2) #sensor tgl valid
cv.rectangle(kel5, pt1=(290,700), pt2=(365,740), color=(0,255,255), thickness=2) #sensor tulisan D-IV

plt.imshow(cv.cvtColor(kel5, cv.COLOR_BGR2RGB))
plt.show()
```
