````python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow # for image display
from skimage import io
from skimage import transform
from PIL import Image, ImageDraw
import matplotlib.pylab as plt
````

````python
# Membuat list untuk menyimpan url dari beberapa image
urls = ["https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
        "https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg",
        "https://i.pinimg.com/736x/f9/5f/43/f95f43dbd59cedfc8bc99f42c8a5c80a.jpg"]
# baca dan tampilkan image
# loop pada tiap url image, beberapa image dapat disimpan pada list
for url in urls:
    image = io.imread(url)                           #read image
    image = cv.resize(image, (0,0), fx=0.5, fy=0.5)  #resize image to half size
    image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)   #convert color to RGB
    final_frame = cv.hconcat((image, image_2))       #concatenate image
    cv2_imshow(final_frame)                          #show image
    print('\n')
````

````python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("resolusi image: tinggi x lebar = ",tinggi," x ",lebar)
cv2_imshow(image_2)
````

````python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

#membuat garis horizontal ditengah image
for y in range (lebar):
  image_3[int((tinggi)/2),y] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
````

4. Perhatikan potongan kode progam berikut :


```
for y in range (lebar):
  image_3[int((tinggi)/2),y] = [255,255,255]
```


Apakah kegunaan kode [255,255,255] ? Jelaskan!
> Pada pemrosesan gambar menggunakan OpenCV, piksel direpresentasikan sebagai nilai numerik untuk setiap saluran warnanya. Untuk gambar RGB (Red, Green, Blue) 8-bit, setiap saluran memiliki rentang nilai dari 0 hingga 255.

> [255,255,255] mewakili nilai maksimum untuk setiap saluran warna (Merah=255, Hijau=255, Biru=255). Kombinasi ini menghasilkan warna putih.
Dalam kode image_3[int((tinggi)/2),y] = [255,255,255], ini berarti bahwa setiap piksel pada baris tengah gambar (int((tinggi)/2)) akan diatur menjadi warna putih saat iterasi y melintasi lebar gambar. Dengan demikian, kode ini berfungsi untuk menggambar garis horizontal berwarna putih tepat di tengah gambar.

5. Jelaskan keterkaitan antara pixel dan juga resolusi gambar yang tinggi ataupun rendah!
> - Piksel: Piksel adalah unit dasar terkecil dari sebuah gambar digital. Setiap gambar digital yang Anda lihat di layar atau cetak terdiri dari jutaan piksel kecil yang tersusun dalam bentuk grid. Setiap piksel menyimpan informasi warna tertentu.
> - Resolusi Gambar: Resolusi gambar mengacu pada jumlah total piksel yang membentuk sebuah gambar. Ini biasanya dinyatakan sebagai lebar x tinggi (misalnya, 1920x1080 piksel).
> - Resolusi Tinggi: Gambar dengan resolusi tinggi memiliki jumlah piksel yang besar. Ini berarti gambar tersebut dapat menangkap dan menampilkan lebih banyak detail. Hasilnya adalah gambar yang lebih jernih, tajam, dan dapat diperbesar tanpa kehilangan kualitas yang signifikan atau terlihat 'pecah' (pikselasi).
> - Resolusi Rendah: Gambar dengan resolusi rendah memiliki jumlah piksel yang lebih sedikit. Ini berarti ada lebih sedikit detail yang dapat ditampilkan. Gambar akan terlihat kurang tajam, mungkin buram, dan jika diperbesar, piksel-piksel individual akan terlihat jelas (pikselasi), mengurangi kualitas gambar.
> - Keterkaitan: Piksel adalah blok bangunan dasar, dan resolusi adalah jumlah blok bangunan tersebut. Semakin banyak piksel (resolusi lebih tinggi), semakin banyak detail yang dapat disimpan dan ditampilkan, menghasilkan kualitas gambar yang lebih baik. Sebaliknya, semakin sedikit piksel (resolusi lebih rendah), semakin sedikit detail dan kualitas gambar pun menurun.



````python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

#membuat garis vertikal ditengah image
for x in range (tinggi):
  image_3[x, int((lebar)/2)] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
````

````python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# membuat garis diagonal menyilang pada image
for x in range(tinggi):
    # Diagonal dari kiri atas ke kanan bawah
    y1 = int(x * (lebar - 1) / (tinggi - 1))
    if y1 < lebar: # Pastikan koordinat y tidak melebihi lebar gambar
        image_3[x, y1] = [255,255,255]

    # Diagonal dari kanan atas ke kiri bawah
    y2 = int((tinggi - 1 - x) * (lebar - 1) / (tinggi - 1))
    if y2 >= 0 and y2 < lebar: # Pastikan koordinat y tidak melebihi lebar gambar
        image_3[x, y2] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
````

````python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

#membuat garis horizontal di tengah image dengan panjang yang bisa diatur
# Tentukan titik awal dan akhir garis horizontal
start_length = int(lebar * 0.25) # Mulai dari 25% lebar gambar
end_length = int(lebar * 0.75)   # Berakhir di 75% lebar gambar

for y in range (start_length, end_length):
  image_3[int((tinggi)/2),y] = [255,255,255] # Garis di tengah tinggi gambar

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
````

````python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

#membuat kotak dengan kumpulan pixel putih di sembarang tempat dengan ukuran bisa diatur
# Tentukan posisi sudut kiri atas kotak
start_x = 150  # Koordinat X awal
start_y = 150  # Koordinat Y awal

# Tentukan ukuran kotak
box_width = 100
box_height = 80

# Pastikan kotak tidak keluar dari batas gambar
end_x = min(start_x + box_height, tinggi)
end_y = min(start_y + box_width, lebar)

for x in range(start_x, end_x):
  for y in range(start_y, end_y):
    image_3[x, y] = [255,255,255] # Set pixel menjadi putih

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
````
